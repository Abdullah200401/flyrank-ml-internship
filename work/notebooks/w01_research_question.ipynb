## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Abdullah200401/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

display(df.head())
print("Average search volume:", df["search_volume"].mean())
print("Average CTR:", df["ctr"].mean())
print("Average engagement rate:", df["engagement_rate"].mean())
```

The starter dataset contains 30,000 rows and 44 columns. The average search volume is 158.88, while the average CTR is 0.51 and the average engagement rate is 2.53. These numbers show that the dataset contains measurable search and engagement signals at the page level. This makes it reasonable to investigate whether these signals can help prioritize pages for possible content refresh, while recognizing that these averages alone do not prove that a refresh will improve performance.
