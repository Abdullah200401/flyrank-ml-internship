```python
# Section 3: Verify grain, counts, missing values, and window

import pandas as pd

df_query = query_table.to_pandas()

# 1. Grain check: one row should represent one content-query pair
print("=== 1. Grain check ===")
print("Total rows:", len(df_query))
print(
    "Duplicate content-query pairs:",
    df_query.duplicated(
        subset=["content_hash_id", "query_hash_id", "window_start", "window_end"]
    ).sum()
)

# 2. Counts and date window
print("\n=== 2. Counts and window ===")
print("Unique content items:", df_query["content_hash_id"].nunique())
print("Unique queries:", df_query["query_hash_id"].nunique())
print("Window start:", df_query["window_start"].min())
print("Window end:", df_query["window_end"].max())

# 3. Availability check using IS TRUE equivalent
print("\n=== 3. Availability ===")
available = df_query[df_query["impressions_90d"].notna()]
print("Rows with impressions_90d available:", len(available))
print("Rows surviving availability check:", len(available))
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

Data limits:

- The data cannot tell us the true business impact of refreshing a page because we do not observe a direct business outcome such as revenue or conversions.
- The 90-day query data gives a historical search-performance view, but it does not prove that a future refresh will cause the observed changes.
- Some performance data may only be available for clients with the relevant GSC or GA4 access.
- The time windows can overlap, so rows should not be treated as fully independent observations.
- June 2026 should be treated as a sealed outcome/test period rather than used to develop the ranking logic.
