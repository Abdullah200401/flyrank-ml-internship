## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

A fixed rule could use one threshold, such as selecting every page with CTR below a certain value. However, refresh opportunity may depend on several signals at the same time. For example, a page with low CTR but very little search demand may be less important than a page with meaningful search demand, weak average position, declining trend, and low engagement. A scoring or ranking approach can combine these signals and produce a more flexible prioritization than a single if-statement. The model would support human review rather than automatically deciding which pages must be refreshed.
